# Quantum Hello World

This notebook demonstrates a basic quantum circuit — the **Bell State** — using Qiskit.

A Bell state is a maximally entangled two-qubit state:

$$
|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}
$$

When measured, both qubits will always be in the same state (both 0 or both 1),
with equal probability.

````python
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram
from qiskit_aer import AerSimulator

print(f"Qiskit version: {qiskit.__version__ if 'qiskit' in dir() else 'loading...'}")
````

## Build the circuit

Create a 2-qubit circuit, apply a Hadamard gate to qubit 0, then a
CNOT (controlled-NOT) from qubit 0 to qubit 1. Finally measure both qubits.

The Hadamard puts qubit 0 into a superposition, and the CNOT entangles
qubit 1 with it — producing the Bell state $|\Phi^+\rangle$.

````python
qc = QuantumCircuit(2, 2)
qc.h(0)        # Hadamard gate on qubit 0
qc.cx(0, 1)    # CNOT: qubit 0 controls qubit 1
qc.measure([0, 1], [0, 1])  # measure both qubits

print(qc.draw())
````

## Run on simulator

Execute the circuit on the local **AerSimulator** (the default execution mode).
The result should show approximately equal counts for `00` and `11`.

> **Tip**: Switch to **Real Hardware** in your Quantum Settings to run on
> an actual IBM quantum computer. The same circuit works — just slower,
> and on real hardware the counts will include noise.

````python
backend = AerSimulator()
result = backend.run(qc, shots=1024).result()
counts = result.get_counts()

print("Counts:", counts)

# Verify entanglement: both qubits always agree
if abs(counts.get('00', 0) - counts.get('11', 0)) < 150:
    print("\n✓ Bell state verified — qubits are entangled!")
else:
    print("\n⚠ Unexpected distribution — check circuit.")
````

````python
# Visualise the histogram
fig = plot_histogram(counts)
display(fig)
````
